# This cell may not be required as the coordinate system used by the the datasets is already EPSG:4326

```python
points = gpd.GeoDataFrame(
    grainotheque,  # the original dataframe
    geometry=gpd.points_from_xy(grainotheque["lon"], grainotheque["lat"]),
    crs="EPSG:4326",              # what the numbers ARE: UTM zone 10N
)
web = points.to_crs("EPSG:4326")  # what web maps want: longitude and latitude
web.geometry.head()
```

```python
#Now we may keep only some of the columns in grainotheque
#geometry is the column that contains the coordinates of the points, and it is needed for the GeoJSON file. 
keep = web[["name","adresse", "code.postal", "site.internet","resources","dispo","document.requis", "geometry"]]
keep.to_file("../data/grainotheque_mtl.geojson", driver="GeoJSON", COORDINATE_PRECISION=6)
print(round(os.path.getsize("../data/grainotheque_mtl.geojson") / 1_000_000, 1), "MB")
```

# Check Cell to see if the map will write the features correctly

```python
# ---- settings: specify/define what are the elements to check ----
FILE = "../data/grainotheque_mtl.geojson"
WEST, EAST, SOUTH, NORTH = -74, -73.3, 45.4, 45.8   # a box around the area, in degrees
EXPECTED = 32           # how many features you expect (rows after your filters)
POPUP_FIELDS = ["name","adresse", "code.postal", "site.internet","resources","dispo","document.requis"]  # which columns you want to show in the popup
# ----------------------------------------------------

back = gpd.read_file(FILE)
problems = []
if len(back) != EXPECTED:
    problems.append(f"{len(back)} features, but you expected {EXPECTED}.")
if back.crs is None or back.crs.to_epsg() != 4326:
    problems.append(f"The file is in {back.crs}, not longitude and latitude (EPSG:4326).")
minx, miny, maxx, maxy = back.total_bounds
if not (WEST <= minx and maxx <= EAST and SOUTH <= miny and maxy <= NORTH):
    problems.append(f"Some features are outside your box: x {minx:.3f} to {maxx:.3f}, y {miny:.3f} to {maxy:.3f}.")
missing = [f for f in POPUP_FIELDS if f not in back.columns]
if missing:
    problems.append(f"Missing popup fields: {missing}.")
mb = os.path.getsize(FILE) / 1_000_000
if mb > 5:
    problems.append(f"The file is {mb:.1f} MB; over about 5 MB the page gets slow. Filter or simplify.")
print("Geometry types:", sorted(back.geom_type.unique()))
print("All checks passed" if not problems else "Not yet:\n- " + "\n- ".join(problems))
```

```python
#Here we create a function that whose output is a featureand it will do that for any row
def row_to_feature(row):
    """Take one row of the sewer table and return one GeoJSON feature."""
    return  {
    "type": "Feature",
    "geometry": {
        "type": "Point",
        "coordinates": [float(row["lon"]),float(row["lat"])],        # longitude first, then latitude
    },
    "properties": {
        "arrondissement": row["arrondissement"],
        "adresse ": row["adresse"],
        "code_postal": row["code.postal"],                  # we assign each property a name
        "site_internet": row["site.internet"],
        "telephone_general ": row["telephone.general"],
        "resources": row["resources"],
        "dispo": row["dispo"],
        "document_requis": row["document.requis"],
        "description": row["description"]


    },
}
print(json.dumps(feature, indent=2))
```

```python
features = []
for _, row in grainotheque.iterrows():
    features.append(row_to_feature(row)) #here this function creates alist with each newly created features
                          # add row_to_feature(row) to the list

collection = {"type": "FeatureCollection", "features": features}

out_path = "../data/polliresources.geojson"
with open(out_path, "w", encoding="utf-8") as f:
    json.dump(collection, f, separators=(",", ":"), allow_nan=False)
    # separators: no spaces, a smaller file.
    # allow_nan=False: stop with an error if any value is NaN (an empty cell).
    # Browsers cannot read NaN in JSON, so without this line the file writes
    # fine and the map fails later with a confusing message. Fail here instead.

import os
print(len(features), "features written to", out_path)
print(round(os.path.getsize(out_path) / 1_000_000, 1), "MB")
```
