```python
import numpy as np
import geopandas as gpd
from shapely.geometry import Polygon

def make_hex_grid(gdf, side):
    """Flat-top hexagons covering gdf's bounds; coordinates must be in metres."""
    minx, miny, maxx, maxy = gdf.total_bounds
    radius = side
    dx = 1.5 * radius
    dy = np.sqrt(3) * radius

    hexagons = []
    x_centers = np.arange(minx - 2 * radius, maxx + 2 * radius, dx)

    for col, x in enumerate(x_centers):
        y_centers = np.arange(miny - 2 * radius, maxy + 2 * radius, dy)
        offset = dy / 2 if col % 2 else 0

        for y in y_centers:
            center_y = y + offset
            vertices = [
                (
                    x + radius * np.cos(np.deg2rad(angle)),
                    center_y + radius * np.sin(np.deg2rad(angle)),
                )
                for angle in range(0, 360, 60)
            ]
            hexagons.append(Polygon(vertices))

    return gpd.GeoDataFrame(
        {"cell_id": range(len(hexagons))},
        geometry=hexagons,
        crs=gdf.crs,
    )

grid = make_hex_grid(web, side=500)
print(len(grid), "hexagons")
```

```python
joined = gpd.sjoin(web, grid, predicate="intersects")       # which square does each pipe touch?
joined = joined[~joined.index.duplicated()]             # a pipe exactly on an edge touches two squares: keep it once

summary = joined.groupby("cell_id").agg(
    original_state=("original_condition_surface", "count"),        # how many pipes in the square
    in_very_good_state=("is_original_condition_surface", "sum"),   # how many of them are combined: add up the True values
).reset_index()
summary["very_good_state_pct"] = (100 * summary["in_very_good_state"] / summary["original_state"]).round(0)

cells = grid.merge(summary, on="cell_id")   # keeps only squares that have pipes
print(len(cells), "hexagons have ruelle vertes")
print("Ruelle vertes counted:", cells["in_very_good_state"].sum(), "of", len(web))
cells.head()
cells.iloc[80]
```

```python
print(cells["very_good_state_pct"].describe().round(1))
print()
print("hexagons with no ruelles in good state:", (cells["very_good_state_pct"] == 0).sum())
print("hexagons that have at least half of the ruelles in very good states:", (cells["very_good_state_pct"] >= 50).sum())
```

```python
web_cells = cells[["cell_id","original_state","in_very_good_state","very_good_state_pct", "geometry"]].to_crs("EPSG:4326")
web_cells.to_file("../data/ruelle_grid.geojson", driver="GeoJSON", COORDINATE_PRECISION=6)

combined_pipes = web[web["is_original_condition_surface"]].to_crs("EPSG:4326")[["name", "original_condition_surface", "original_arrondissement", "geometry"]]
combined_pipes.to_file("../data/in_very_good_state_ruelles.geojson", driver="GeoJSON", COORDINATE_PRECISION=6)

for f in ["../data/ruelle_grid.geojson", "../data/in_very_good_state_ruelles.geojson"]:
    print(f, round(os.path.getsize(f) / 1_000_000, 2), "MB")
```
